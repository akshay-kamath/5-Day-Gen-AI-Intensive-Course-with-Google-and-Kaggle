````python
# Import Google's retry utility to automatically handle retry logic for transient API errors
from google.api_core import retry

# Define a lambda function to check if an error is retriable
# This targets rate limiting (429) or service unavailable (503) errors
is_retriable = lambda e: (
    isinstance(e, genai.errors.APIError) and e.code in {429, 503}
)

# Apply retry logic to Gemini's generate_content method
# This ensures that if a retriable error occurs, the call will be retried automatically
genai.models.Models.generate_content = retry.Retry(
    predicate=is_retriable
)(genai.models.Models.generate_content)
````

# Structured Output/JSON Mode/Controlled Generation

* Daily Wellness Plan Generation

    * The agent generates a Daily Wellness Plan with a focus on improving mental health, particularly targeting areas like reducing anxiety, boosting mood= or improving focus.
    
    * The generated plan is returned in structured JSON format, making it easy to process programmatically or display dynamically in a user interface.

* How it works:

    * Input Focus Area: The user specifies the area of focus (e.g., "reducing anxiety" or "improving focus").
    AI Content Generation: The system sends a prompt to the Gemini 2.0 Flash model, asking it to generate a daily wellness plan based on the focus area.
    
    * Structured Output: The output is formatted as a structured JSON object, which includes:
    
    * Daily Goal: A concise goal related to the focus area (e.g., "Reduce anxious overthinking").
    
    * Action Steps: A list of 2-3 specific actions to achieve the goal.
    
    * Mantra: A positive affirmation to encourage the user throughout the day.

    Example JSON Output for a Daily Wellness Plan:
    
    ```json
    {
      "daily_goal": "Reduce anxious overthinking",
      "action_steps": [
        "Practice 5-minute breathing before work",
        "Challenge negative thoughts with evidence",
        "Reflect on 1 thing I did well today"
      ],
      "mantra": "I am steady, strong, and supported by the universe."
    }
    ```

````python
import json
from typing import Dict, List, Optional

# Function to generate a daily wellness plan based on a specific area of focus
def generate_daily_wellness_plan(focus: str) -> Optional[Dict]:
    """
    Generates a daily wellness plan in JSON format focused on a specific area.

    Args:
        focus: The area of focus for the wellness plan (e.g., "reducing anxiety",
               "improving focus", "boosting mood").

    Returns:
        A dictionary representing the wellness plan if successful, otherwise None.
    """
    # Define the response schema in JSON format
    response_schema = {
        "type": "object",
        "properties": {
            "daily_goal": {"type": "string", "description": f"A concise goal for the day related to {focus}"},
            "action_steps": {
                "type": "array",
                "items": {"type": "string", "description": "An actionable step to achieve the daily goal"},
                "minItems": 2,
                "maxItems": 3,
                "description": "2-3 specific actions to take today"
            },
            "mantra": {"type": "string", "description": "A short, positive affirmation related to the daily goal"}
        },
        "required": ["daily_goal", "action_steps", "mantra"],
        "description": f"A daily wellness plan focused on {focus}"
    }

    prompt = f"""
    Generate a daily wellness plan focused on {focus}.
    The plan should include a daily goal, 2-3 actionable steps and a relevant mantra.
    Please format your response as a JSON object according to the following schema:
    ```json
    {json.dumps(response_schema, indent=2)}
    ```
    """

    try:
        response =  client.models.generate_content(
            model="gemini-2.0-flash",
            contents=prompt,
            config=types.GenerateContentConfig(
            temperature=0.1,
            response_mime_type="application/json",
            response_schema=response_schema
            )

        )
        if response.text:
            return json.loads(response.text)
        else:
            print("Received an empty response from the model.")
            return None
    except json.JSONDecodeError as e:
        print(f"Error decoding JSON: {e}")
        print(f"Raw response: {response.text}")
        return None
    except Exception as e:
        print(f"An error occurred during content generation: {e}")
        return None

# Example usage of the function to generate a wellness plan for a specific focus area
focus_area = "reducing anxious overthinking"
wellness_plan = generate_daily_wellness_plan(focus_area)

# If a valid wellness plan is generated, print it in a pretty JSON format
if wellness_plan:
    print(json.dumps(wellness_plan, indent=4))
````

# Few Shot Prompting


One of the most powerful techniques we use in ChetnaShakti is Few Shot Prompting, where we provide the model with a set of carefully constructed examples to guide it in generating empathetic, context aware mental health responses.

**Use Case: CBT Reframing with Compassion**

* This component helps users reframe intrusive or negative thoughts using CBT techniques such as:

    * Identifying thought distortions (e.g. catastrophizing, overgeneralization)
    
    * Challenging irrational beliefs
    
    * Replacing them with grounded, compassionate self talk

**Prompt Design**

We feed the Gemini model with 2–3 real life examples of negative thoughts and how a CBT coach would reframe them. When a user submits their own concern (e.g., “I feel worthless and want to die”), the model uses this context to output a structured JSON with:

```json
{
  "original_thought": "...",
  "thought_distortion": "...",
  "challenge": "...",
  "balanced_thought": "..."
}
```
* This structured output is both emotionally nuanced and machine readable making it suitable for integration into mental health apps or daily journals.

**Behavioral Coaching via Habit Agent**

**Use Case: Atomic Habit Planning**

* Using the "habit loop" framework (cue → routine → reward), we designed a behavioral coach that helps users build sustainable habits aligned with personal goals like:

    * Reducing screen time
    
    * Improving focus
    
    * Building a mindfulness routine

* Output Example:
```json

{
  "habit_goal": "Improve focus and reduce screen time",
  "habit_loop": {
    "cue": "Feeling distracted or bored",
    "routine": "Switch to a focus playlist and close social apps",
    "reward": "Sense of accomplishment + track streak"
  },
  "tracking_tip": "Use a habit tracker with emoji reactions",
  "long_term_routine_suggestion": "Schedule 2 daily no-phone focus blocks"
}
```

````python
# User input for CBT reframing
user_input = "I feel life is too harsh to me.I feel worthless and want to die."

def get_cbt_response(user_input, client, model_name="gemini-2.0-flash"):
    """
    Uses Few-Shot Prompting to generate a CBT-style reframe for a given negative thought.
 
    Given a user_input string, returns a CBT-style reframe using Gemini model.

    Parameters:
        user_input (str): The negative thought to be reframed.
        client: The Gemini client instance.
        model_name (str): Gemini model name to use.

    Returns:
        str: A JSON-formatted CBT response including thought distortion, challenge and balanced thought.
    """    
    cbt_prompt = f"""
    You are a compassionate CBT (Cognitive Behavioral Therapy) coach.
    
    Your goal is to help users recognize and reframe negative thoughts using 4 steps:
    1. Identify the thought
    2. Detect the distortion
    3. Challenge the thought
    4. Replace it with a balanced thought
    
    Examples:
    
    User: “I always mess things up.”
    Coach:
    - Thought distortion: Overgeneralization
    - Challenge: “Always? Can you think of a time you succeeded?”
    - Balanced thought: “I’ve made mistakes, but I’ve also succeeded before.”
    
    User: “Nobody likes me.”
    Coach:
    - Thought distortion: Mind reading
    - Challenge: “How do you know what others think?”
    - Balanced thought: “Some people might not connect with me, but many do.”

    Now respond to this user input:
    
    User: “{user_input}”

    Return output in following JSON format for example:
      "original_thought": "I feel life is too harsh to me. I feel worthless and want to die.",
      "thought_distortion": "Catastrophizing and labeling",
      "challenge": "Life can feel overwhelming, but are there moments, even small ones, where things felt okay or you felt valued? Feeling worthless doesn't mean you truly are.",
      "balanced_thought": "Life is difficult right now, but that doesn't mean it always will be. I have worth even if I can't feel it in this moment. I can take things one step at a time and seek support."
    """

    response = client.models.generate_content(
        model="gemini-2.0-flash",
        contents=cbt_prompt)
    return response.text

# Generate the CBT reframe for the input
user_input = "I feel life is too harsh to me. I feel worthless and want to die."
cbt_reply = get_cbt_response(user_input, client)
print(cbt_reply)
````

````python
def generate_habit_plan(goal: str) -> Dict:
    """
    Generates a personalized habit loop (cue, routine, reward) based on user's goal using few-shot prompting.

    Parameters:
        goal (str): The user’s goal for habit formation.

    Returns:
        dict: A structured JSON containing habit_goal, cue-routine-reward loop, tracking tips and suggestions.
    """
    # Prompt provides structure for atomic habits model + JSON output
    prompt = f"""
    You are a behavioral coach helping users build atomic habits.

    Based on the goal: "{goal}", generate a simple habit loop.

    Return the response as a JSON object:
    {{
      "habit_goal": "...",
      "habit_loop": {{
        "cue": "...",
        "routine": "...",
        "reward": "..."
      }},
      "tracking_tip": "...",
      "long_term_routine_suggestion": "..."
    }}
    """

    response = client.models.generate_content(
        model="gemini-2.0-flash",
        contents=prompt,
        config=types.GenerateContentConfig(
            temperature=0.2,
            response_mime_type="application/json"
        )
    )

    try:
        return json.loads(response.text)
    except Exception as e:
        print("Error parsing HabitAgent response:", e)
        return {}

# Generate habit plan for a common goal
habit_response = generate_habit_plan("Improve focus and reduce screen time")
print(json.dumps(habit_response, indent=2))
````

````python
#  Helper function to download images 
def download_image(url, save_path):
    try:
        response = httpx.get(url, follow_redirects=True, timeout=20.0)
        response.raise_for_status() # Raise an exception for bad status codes
        path = pathlib.Path(save_path)
        path.write_bytes(response.content)
        print(f"Downloaded image to {save_path}")
        return path
    except httpx.RequestError as e:
        print(f"Error downloading {url}: {e}")
        return None
    except Exception as e:
         print(f"An error occurred during download: {e}")
         return None

#Example 1: Processing Scanned Scripture
def process_scripture_image(image_path: str, user_context: str = "seeking general understanding"):
    """
    Analyzes an image of a Sanskrit scripture verse.

    Args:
        image_path: Path to the image file.
        user_context: Brief description of why the user uploaded this (e.g., seeking strength, understanding detachment).

    Returns:
        String containing the analysis from the model, or an error message.
    """
    if not client:
        return "Error: Model not initialized."
    if not pathlib.Path(image_path).exists():
        return f"Error: Image file not found at {image_path}"

    print(f"\n Processing Scripture Image: {image_path}")
    try:
        img = PIL.Image.open(image_path)

        prompt = f"""Analyze this image suspected to contain a Sanskrit verse from a scripture like the Bhagavad Gita or Durga Saptashati. The user is currently {user_context}.

Please perform the following steps and structure your response clearly using Markdown:
1.  **Transcription:** Extract the Sanskrit text (in Devanagari script) as accurately as possible. If text is unclear, note that.
2.  **Transliteration:** Provide the standard IAST transliteration of the extracted Sanskrit text.
3.  **Translation:** Provide a clear English translation of the verse.
4.  **Source Identification (Optional):** If possible, identify the likely source (e.g., Bhagavad Gita Chapter X, Verse Y, or Durga Saptashati chapter/context). State if unsure.
5.  **Contextual Relevance:** Briefly explain the meaning or significance of this verse, especially considering the user's context of '{user_context}'.
"""

        contents = [prompt, img]
        response = client.models.generate_content(model='gemini-2.0-flash',contents=contents)
        return response.text

    except FileNotFoundError:
         return f"Error: Image file not found at {image_path}"
    except Exception as e:
        print(f"Error processing scripture image: {e}")
        return f"An error occurred during image processing: {e}"

gita_verse_url = "https://upload.wikimedia.org/wikipedia/commons/5/57/6th_Chapter%2C_verse_1%2C_Bhagavad_Gita%2C_Sanskrit%2C_Devanagari_script.jpg" 
gita_image_path = download_image(gita_verse_url, "gita_verse_example.jpg")

if gita_image_path:
    display(Image(str(gita_image_path), width=400))
    analysis = process_scripture_image(str(gita_image_path), user_context="feeling overwhelmed by duties")
    display(Markdown(analysis))
else:
    print("Skipping scripture analysis due to download error.")
````

````python

def upload_and_process_video(file_path: str) -> Optional[types.File]:
    """Uploads a video file and waits for it to be processed."""
    if not client:
        print("Error: Gemini client not initialized.")
        return None
    if not pathlib.Path(file_path).exists():
        print(f"Error: Video file not found at {file_path}")
        return None

    print(f"Uploading video '{file_path}' to the File API...")
    try:
        video_file = client.files.upload(file=file_path)
        print(f"Upload complete. File Name: {video_file.name}")
    except Exception as e:
        print(f"Error during video upload: {e}")
        return None

    print("Waiting for video processing...")
    while video_file.state.name == "PROCESSING":
        print('.', end='', flush=True)
        time.sleep(5) # Check every 5 seconds
        try:
            video_file = client.files.get(name=video_file.name)
        except Exception as e:
             print(f"\nError getting file status: {e}")
             # Decide how to handle - retry, fail, etc.
             time.sleep(10) # Wait longer before potentially retrying

    if video_file.state.name == "FAILED":
        print(f"\nVideo processing failed: {video_file.state.name}")
        try:
            # Attempt to delete the failed file resource
            client.files.delete(name=video_file.name)
            print(f"Cleaned up failed file resource: {video_file.name}")
        except Exception as e_del:
            print(f"Warning: Could not delete failed file resource {video_file.name}: {e_del}")
        return None
    elif video_file.state.name == "ACTIVE":
        print(f"\nVideo processing complete. URI: {video_file.uri}")
        return video_file
    else:
        print(f"\nUnexpected video file state: {video_file.state.name}")
        return None
````

````python
# Wellness Video Analysis Function
#This function takes the processed video file and uses a tailored prompt.

from typing import Optional 

def analyze_wellness_video(video_file: types.File, user_prompt: str = "Analyze my spoken journal entry for emotional state and suggest relevant wellness practices.") -> str:
    """
    Analyzes a wellness-related video (e.g., spoken journal, check-in).

    Args:
        video_file: The processed video File object from the File API.
        user_prompt: Optional text from the user explaining the video's context.

    Returns:
        String containing the analysis from the model, or an error message.
    """
    if not client:
        return "Error: Model not initialized."
    if not video_file or video_file.state.name != "ACTIVE":
        return "Error: Invalid or unprocessed video file provided."

    print(f"\n Analyzing Wellness Video: {video_file.name} ")

    disclaimer = (
        "\n\n*Disclaimer: AI analysis of emotion and sentiment is experimental and may not be fully accurate. "
        "This is not a substitute for professional medical or psychological diagnosis or advice. "
        "Please consult with a qualified professional for personal health concerns.*"
    )

    #  Prompt for Wellness Analysis 
    analysis_prompt = f"""You are an AI assistant analyzing a user's video for wellness insights. The user provided this context: "{user_prompt}"

Please analyze the video considering the following aspects and structure your response using Markdown:

1.  **Transcription:** Provide a summary or key phrases transcribed from the spoken content. If audio is unclear or absent, state that.
2.  **Emotional Analysis (Multimodal):**
    *   **Vocal Tone:** Describe the perceived tone of voice (e.g., calm, anxious, energetic, fatigued, flat, distressed).
    *   **Facial Expression (if visible):** Describe any discernible facial expressions related to emotion (e.g., smiling, frowning, showing tension, neutral). If the face is not clearly visible, state that.
    *   **Overall Sentiment:** Based on transcription, tone, and expression (where available), assess the likely overall emotional sentiment (e.g., positive, negative, neutral, mixed, leaning towards frustration, expressing hope, showing vulnerability).
3.  **Key Themes/Keywords:** Identify the main topics or recurring keywords mentioned in the transcription (e.g., "overwhelmed," "grateful," "stuck," "relationship," "sleep," "purpose").
4.  **Wellness Suggestions (Grounded in Analysis):** Based on the combined analysis above (sentiment, themes), provide 1-2 relevant suggestions drawing from these areas:
    *   **CBT:** If negative thought patterns seem present (based on transcription/sentiment), suggest a simple reframing question or a focus point (e.g., "Consider asking: 'Is this thought 100% true?'", "Notice if 'all-or-nothing' thinking is present.").
    *   **Habit:** If the user mentions goals, routines, or struggles with action, suggest a tiny habit idea (e.g., "Try linking [desired action] to an existing cue like brushing teeth," "Focus on just 2 minutes of [activity]").
    *   **Spiritual:** If themes of meaning, peace, strength, or specific spiritual language arise, suggest a relevant concept, mantra, or reflection point (e.g., "Reflect on the concept of acceptance," "Consider the mantra 'Om Shanti' for peace," "Explore the Durga Saptashati verse 'Ya Devi Sarva Bhuteshu...' for inner strength if seeking resilience").
    *   **Mindfulness/Somatic:** If stress or overwhelm is evident, suggest a brief grounding or breathing exercise (e.g., "Try a 1-minute body scan," "Practice 4-7-8 breathing").

**Important:** Clearly state the basis for each suggestion (e.g., "Based on the theme of feeling overwhelmed, consider..."). Acknowledge limitations if audio/video quality hinders analysis.
"""

    try:
        # Send the prompt and the video file to the model
        contents = [analysis_prompt, video_file]
        response = client.models.generate_content(model='gemini-2.0-flash',contents=contents)

        # Append the disclaimer to the model's response text
        return response.text + disclaimer

    except Exception as e:
        print(f"Error generating content from video: {e}")
        if "quota" in str(e).lower():
             return "Sorry, I couldn't analyze the video due to current usage limits. Please try again later." + disclaimer
        return f"An error occurred during video analysis: {e}" + disclaimer
````

# State Definition WellnessState

    * next_agent: String denoting which agent should respond next

````python
class WellnessState(TypedDict):
    """Represents the state of the wellness assistant conversation."""
    messages: Annotated[List[BaseMessage], add_messages]
    user_input: Optional[str] # Last user input
    agent_outcome: Optional[str] # Raw output from the last agent
    next_agent: Optional[Literal["CBT", "Habit", "Spiritual","Information", "End", "Clarify"]] # Which agent to run next
````

````python
# --- 4. Agent Core Logic Functions 
def get_cbt_response(user_input: str, client: genai.GenerativeModel) -> str:
    """Given user_input, returns a CBT-style reframe using Gemini."""
    cbt_prompt = f"""
    You are a compassionate CBT (Cognitive Behavioral Therapy) coach. Your goal is to help users recognize and reframe negative thoughts using 4 steps:
    1. Identify the thought
    2. Detect the distortion
    3. Challenge the thought
    4. Replace it with a balanced thought

    Respond ONLY with the JSON output, no preamble or explanation.

    Example Input: "I always mess things up."
    Example Output:
    ```json
    {{
      "original_thought": "I always mess things up.",
      "thought_distortion": "Overgeneralization",
      "challenge": "Always? Can you think of a time you succeeded?",
      "balanced_thought": "I’ve made mistakes, but I’ve also succeeded before."
    }}
    ```

    Example Input: "Nobody likes me."
    Example Output:
    ```json
    {{
      "original_thought": "Nobody likes me.",
      "thought_distortion": "Mind reading",
      "challenge": "How do you know what others think?",
      "balanced_thought": "Some people might not connect with me, but many do."
    }}
    ```

    Now respond to this user input in the same JSON format:
    User: "{user_input}"
    """
    try:
        # Request JSON output directly if supported well by the model
        response = client.generate_content(
            contents=cbt_prompt,
            generation_config=genai.types.GenerationConfig(
                response_mime_type="application/json",
                temperature=0.3 # Allowing for some creativity in reframing
            )
        )
        return response.text
    except Exception as e:
        print(f"Error in get_cbt_response LLM call: {e}")
        # Fallback to text generation if JSON fails
        try:
            response = client.generate_content(contents=cbt_prompt)
            # Attempt to extract JSON from the text response
            match = re.search(r'```json\s*(\{.*?\})\s*```', response.text, re.DOTALL)
            if match:
                return match.group(1)
            else:
                # If no JSON block found, return the raw text as a fallback
                 return json.dumps({"error": "Failed to generate structured CBT response.", "raw_response": response.text})
        except Exception as e2:
             print(f"Fallback error in get_cbt_response: {e2}")
             return json.dumps({"error": "Failed to generate any CBT response."})

````

````python
def generate_habit_plan(goal: str, client: genai.GenerativeModel) -> str:
    """Generates an atomic habit loop for the given goal."""
    prompt = f"""
    You are a behavioral coach helping users build atomic habits.
    Based on the goal: "{goal}", generate a simple habit loop.

    Return ONLY the response as a JSON object, no preamble or explanation:
    ```json
    {{
      "habit_goal": "...",
      "habit_loop": {{
        "cue": "...",
        "routine": "...",
        "reward": "..."
      }},
      "tracking_tip": "...",
      "long_term_routine_suggestion": "..."
    }}
    ```
    """
    try:
        response = client.generate_content(
            contents=prompt,
            generation_config=genai.types.GenerationConfig(
                response_mime_type="application/json",
                temperature=0.2
            )
        )
        return response.text
    except Exception as e:
        print(f"Error in generate_habit_plan LLM call: {e}")
        # Fallback
        try:
            response = client.generate_content(contents=prompt)
            match = re.search(r'```json\s*(\{.*?\})\s*```', response.text, re.DOTALL)
            if match:
                return match.group(1)
            else:
                return json.dumps({"error": "Failed to generate structured habit plan.", "raw_response": response.text})
        except Exception as e2:
            print(f"Fallback error in generate_habit_plan: {e2}")
            return json.dumps({"error": "Failed to generate any habit plan response."})
````

````python

def cbt_agent_node(state: WellnessState) -> WellnessState:
    """Handles CBT tasks."""
    print(">> Calling CBT Agent...")
    user_input = state.get("user_input")
    if not user_input:
        return {"messages": [AIMessage(content="I need your input to provide a CBT analysis.")]}

    # Use partial to pass the client to the function
    cbt_func = partial(get_cbt_response, client=sdk_client_flash)
    response_json_str = cbt_func(user_input)

    # Prepare a user-facing message and store raw output
    ai_message = AIMessage(
        content="Here's a CBT perspective on that thought:",
        additional_kwargs={"raw_output": response_json_str, "agent_name": "CBT"} # Store raw output
    )
    return {"messages": [ai_message], "agent_outcome": response_json_str}


def habit_agent_node(state: WellnessState) -> WellnessState:
    """Handles Habit formation tasks."""
    print(">> Calling Habit Agent...")
    user_input = state.get("user_input")
    if not user_input:
         return {"messages": [AIMessage(content="What habit or goal are you focusing on?")]}

    # Use partial to pass the client
    habit_func = partial(generate_habit_plan, client=sdk_client_flash)
    response_json_str = habit_func(user_input) # Assuming user input describes the goal

    ai_message = AIMessage(
        content="Let's build a habit loop for that:",
        additional_kwargs={"raw_output": response_json_str, "agent_name": "Habit"}
    )
    return {"messages": [ai_message], "agent_outcome": response_json_str}


def spiritual_agent_node(state: WellnessState) -> WellnessState:
    """Handles Spiritual guidance tasks, using cached model if available."""
    print(">> Calling Spiritual Agent...")
    user_input = state.get("user_input")
    if not user_input:
        return {"messages": [AIMessage(content="How are you feeling today? What's on your mind?")]}

    response_text = ""
    model_to_use = None

    if spiritual_cache:
        try:
            # Create a model instance FROM the cache
            print(f"Attempting to use spiritual cache: {spiritual_cache.name}")
            model_to_use = genai.GenerativeModel.from_cached_content(cached_content=spiritual_cache)
            print("Successfully created model from cache.")
        except Exception as e:
            print(f"Error creating model from cache: {e}. Falling back to base model.")
            # Fallback to the base pro model if cache loading fails
            model_to_use = sdk_client_pro # Use the pre-initialized base pro model
    else:
        # If cache doesn't exist at all, use the base pro model
        print("Spiritual cache not available. Using base model.")
        model_to_use = sdk_client_pro # Use the pre-initialized base pro model

    # Ensure we have a model instance before proceeding
    if not model_to_use:
         error_msg = "Fatal error: No valid model instance available for Spiritual Agent."
         print(error_msg)
         return {"messages": [AIMessage(content="I'm sorry, there's a configuration issue with the spiritual guidance module.")]}

    # Call the updated generate_spiritual_insight with the chosen model instance
    response_text = generate_spiritual_insight(user_input, client=model_to_use)

    # Prepare the message for the state update
    ai_message = AIMessage(
        content="Reflecting on that, here's some spiritual perspective:",
         additional_kwargs={"raw_output": response_text, "agent_name": "Spiritual"}
    )
    return {"messages": [ai_message], "agent_outcome": response_text}


def information_agent_node(state: WellnessState) -> WellnessState:
    """Handles informational queries using the grounded model."""
    print(">> Calling Information Agent (with grounding)...")
    user_input = state.get("user_input")
    if not user_input:
        return {"messages": [AIMessage(content="What information are you looking for?")]}

    # DISCLAIMER 
    disclaimer = (
        "\n\n*Disclaimer: I am providing general information found via Google Search. "
        "This is not medical or psychological advice. Please consult with a qualified professional "
        "for personal health concerns.*"
    )

    response_text = "Sorry, I couldn't find information on that topic."
    citations_text = ""
    config_with_search = GenerateContentConfig(
            #tools=[types.Tool(google_search=types.GoogleSearch())],
            tools=[Tool(google_search={})]
        )
    try:
        response = client.models.generate_content(model='gemini-2.0-flash',contents=user_input,config=config_with_search)
        response_text = response.text
        print(response_text)
        # --- Process Response and Grounding ---
        markdown_buffer = io.StringIO()
        rc = response.candidates[0]

        # Check if grounding metadata exists before trying to access it
        if hasattr(rc, 'grounding_metadata') and rc.grounding_metadata:
            chunks = rc.grounding_metadata.grounding_chunks
            supports = rc.grounding_metadata.grounding_supports

            # Write the response text segments with citation markers
            markdown_buffer.write("Response:\n\n")
            # Ensure content part exists
            if rc.content and rc.content.parts:
                response_text_content = rc.content.parts[0].text
                # Check if supports exist and process them
                if supports:
                    last_index = 0
                    processed_text = ""
                    for support in supports:
                        start = support.segment.start_index
                        end = support.segment.end_index
                        # Add text before the current support segment
                        processed_text += response_text_content[last_index:start]
                        # Add the support segment text
                        processed_text += response_text_content[start:end]
                        # Add citation markers for this segment
                        for i in support.grounding_chunk_indices:
                            # Ensure index is valid
                            if 0 <= i < len(chunks):
                                processed_text += f"<sup>[{i+1}]</sup>"
                        last_index = end
                    # Add any remaining text after the last support segment
                    processed_text += response_text_content[last_index:]
                    markdown_buffer.write(processed_text)
                else:
                    # If no supports, write the whole text (grounding might not have applied)
                     markdown_buffer.write(response_text_content)
            else:
                 markdown_buffer.write("(No text content found in response)")

            markdown_buffer.write("\n\n") # Add separation

            # Write the citation footnotes if chunks exist
            if chunks:
                markdown_buffer.write("**Sources:**\n\n") # Use Markdown for bold
                for i, chunk in enumerate(chunks, start=1):
                    if hasattr(chunk, 'web') and chunk.web and chunk.web.uri:
                         title = chunk.web.title if chunk.web.title else f"Source {i}"
                         markdown_buffer.write(f"{i}. [{title}]({chunk.web.uri})\n")
                    else:
                         markdown_buffer.write(f"{i}. (Source information unavailable)\n")

            # Get the complete formatted string from the buffer
            formatted_output_markdown = markdown_buffer.getvalue()
            final_content = formatted_output_markdown + disclaimer
            agent_outcome_content = final_content # Update outcome

        else:
            # No grounding metadata, just use the plain text response
            print("No grounding metadata found. Using plain text response.")
            if rc.content and rc.content.parts:
                 final_content = rc.content.parts[0].text + disclaimer
                 agent_outcome_content = final_content # Update outcome
            else:
                 final_content = "(No text content found in response)" + disclaimer
                 agent_outcome_content = final_content # Update outcome


    except Exception as e:
        print(f"Error during grounded generation: {e}")
        # Provide specific error message if possible
        if "quota" in str(e).lower():
             final_content = "Sorry, I'm currently unable to perform searches due to usage limits. Please try again later." + disclaimer
        else:
             final_content = f"Sorry, an error occurred while searching for information." + disclaimer
        agent_outcome_content = final_content # Update outcome with error message

    # --- Create State Update ---
    ai_message = AIMessage(
        content=final_content, # Use the final combined STRING
        additional_kwargs={"agent_name": "Information"}
    )
    return {"messages": [ai_message], "agent_outcome": agent_outcome_content}
     
````

````python
# --- 5. Node Implementations ---

def human_input_node(state: WellnessState) -> WellnessState:
    """Gets input from the user."""
    last_message = state['messages'][-1] if state['messages'] else None

    # Display the last AI message before prompting
    if isinstance(last_message, AIMessage):
        # Get agent name and raw output from additional_kwargs if they exist
        agent_name = last_message.additional_kwargs.get("agent_name")
        raw_output = last_message.additional_kwargs.get("raw_output", "")

        # Check the agent_name to decide how to display
        if agent_name == "CBT":
             print(f"\nAssistant: {last_message.content}") # Print the intro phrase
             display_structured_output("CBT", raw_output) # Display the detailed CBT output
        elif agent_name == "Habit":
             print(f"\nAssistant: {last_message.content}") # Print the intro phrase
             display_structured_output("Habit", raw_output) # Display the detailed Habit output
        elif agent_name == "Spiritual":
             print(f"\nAssistant: {last_message.content}") # Print the intro phrase
             display_spiritual_insight_markdown(raw_output) # Display the detailed Spiritual output
        else:
            # Handle general AI messages that don't have agent_name/raw_output
            print(f"\nAssistant: {last_message.content}")

    # Get user input
    user_input = input("You: ")
    if user_input.lower() in ['q', 'quit', 'exit', 'bye']:
        # Return state update to signal exit
        return {"messages": [HumanMessage(content="exit")], "next_agent": "End"} # Ensure next_agent is set for router

    # Return state update with user message and input
    return {"messages": [HumanMessage(content=user_input)], "user_input": user_input}

def route_to_agent(state: WellnessState) -> Literal["cbt_agent", "habit_agent", "spiritual_agent", "information_agent", "__end__"]:
    """Routes to the appropriate agent based on user input."""
    user_message = state['messages'][-1]
    # if not isinstance(user_message, HumanMessage) or user_message.content == "exit":
    #     return END # End if the last message wasn't from the user or was 'exit'

    #  1. Handle Exit Condition 
    if not isinstance(user_message, HumanMessage) or user_message.content.lower() in ['q', 'quit', 'exit', 'bye', 'goodbye']:
         print("Router decision: Exit")
         return END # End if the last message wasn't from the user or was an exit command    
    user_input_lower = user_message.content.lower()

    #  2. Prioritize Safety Check
    # Check for crisis keywords BEFORE routing to general information or other agents
    crisis_keywords = ["kill myself", "want to die", "suicidal", "hopeless and want to end"]
    if any(keyword in user_input_lower for keyword in crisis_keywords):
         print("Router decision: Crisis detected -> CBT (Safety Priority)")
         # Always route critical safety concerns, potentially to a specialized handler later
         return "cbt_agent"

    #  3. Prepare Prompt for LLM Router 
    prompt = f"""You are a routing agent for a multi-agent wellness chatbot. Your goal is to analyze the user's latest message and determine which specialist agent is best suited to respond.

Choose ONLY ONE of the following options based on the user's primary intent:

1.  **'Information'**: If the user is asking a general knowledge question, seeking definitions, explanations, or facts (e.g., "What is mindfulness?", "Tell me about anxiety symptoms", "Explain cognitive distortions"). Use this for requests for objective information, even if related to wellness topics.
2.  **'CBT'**: If the user expresses negative thoughts, self-criticism, specific worries, anxiety, sadness, difficult emotions, or explicitly wants to challenge their thinking patterns. Focus on personal feelings and cognitive processes.
3.  **'Habit'**: If the user wants to build a new habit, break a bad one, improve routine, focus, discuss productivity, track progress, or needs help with goal-setting related to actions.
4.  **'Spiritual'**: If the user expresses feelings of meaninglessness, seeks deeper connection/purpose, mentions faith, needs strength/resilience/peace in a broader sense, or asks for spiritual guidance/interpretation (but NOT just factual information *about* spirituality - that's 'Information').
5.  **'End'**: If the user explicitly wants to end the conversation (e.g., "that's all for now", "I'm done").

**Prioritization:**
- Prioritize 'CBT' if the user expresses significant distress or negative self-talk, even if other elements are present.
- Distinguish between asking *for* spiritual guidance ('Spiritual') and asking *about* spiritual concepts ('Information').
- If the intent is clearly informational, choose 'Information'.

User message: "{user_message.content}"

Based on the primary intent of the message, which agent should handle this? Respond with ONLY ONE WORD: Information, CBT, Habit, Spiritual, or End."""

    try:
        response = router_llm.invoke(prompt)
        route = response.content.strip().lower()
        route = route.replace('.', '')        
        print(f"LLM Router Raw Output: '{response.content}', Router Decision: '{route}'") # Debugging

        if "information" in route:
            return "information_agent"        
        if "cbt" in route:
            return "cbt_agent"
        elif "habit" in route:
            return "habit_agent"
        elif "spiritual" in route:
            return "spiritual_agent"
        elif "end" in route:
            return END
        else:
            # Default or fallback if unsure - default to CBT as it often deals with general distress
            print("Router fallback to CBT")
            return "cbt_agent"
    except Exception as e:
        print(f"Error in router: {e}")
        # Fallback in case of router error
        return "cbt_agent"

````

# Edges

1. Flow is user driven but guided:

The dialogue flow is controlled by the user’s input, but it is guided by the system. After each response from an agent, the system loops back to the user for further input. This recursive process ensures that the conversation remains relevant and centered on the user’s evolving needs. Each interaction can shape the next steps in the dialogue, creating a dynamic and fluid exchange.

2. Conditional Routing:

The router_llm is responsible for managing the conditional routing between the nodes. It uses intelligent decision-making based on the content of the user's input to determine which agent should respond. For instance, if a user expresses feelings of anxiety, the router might route the conversation to the CBT agent for cognitive reframing. If a user is asking for spiritual guidance, the system will route the query to the Spiritual Agent. This ensures the conversation is always tailored to the user's immediate emotional and intellectual needs.

````python
#  6. Graph Construction 
graph_builder = StateGraph(WellnessState)

# Add nodes
graph_builder.add_node("human_input", human_input_node)
graph_builder.add_node("cbt_agent", cbt_agent_node)
graph_builder.add_node("habit_agent", habit_agent_node)
graph_builder.add_node("spiritual_agent", spiritual_agent_node)
graph_builder.add_node("information_agent", information_agent_node) 

# Define edges
graph_builder.add_edge(START, "human_input") # Start by getting user input


# Conditional routing from the router to specialist agents or end
graph_builder.add_conditional_edges(
    "human_input", # The node whose state we check AFTER it runs
    route_to_agent, # Pass the function directly
    { # Map return values of route_to_agent to subsequent nodes
        "cbt_agent": "cbt_agent",
        "habit_agent": "habit_agent",
        "spiritual_agent": "spiritual_agent",
        "information_agent": "information_agent",        
        END: END, # Map the special END object if route_to_agent returns it
    }
)


# Edges from agents back to human input for the next turn
graph_builder.add_edge("cbt_agent", "human_input")
graph_builder.add_edge("habit_agent", "human_input")
graph_builder.add_edge("spiritual_agent", "human_input")
graph_builder.add_edge("information_agent", "human_input") 
````

````python
@functools.cache
def eval_awareness_enhanced(text: str, cbt_context: Optional[str] = None) -> Tuple[str, Optional[str], Optional[str]]:
    """
    Evaluate one journal entry for emotional insight and awareness.
    Optionally considers previous CBT guidance to check if user is integrating feedback.
    """
    print(f"Evaluating awareness for text: '{text[:50]}...' with CBT context: {cbt_context is not None}")
    # Prepare the prompt with context (if any)
    prompt = AWARENESS_PROMPT_ENHANCED.format(
        text=text,
        cbt_context=cbt_context if cbt_context else "N/A"
    )
    feedback = f"Error evaluating awareness: Unknown error" # Default error message
    score_value = None
    score_name = None

    try:
        # Generate main awareness feedback using the model
        response = client.models.generate_content(
            model="gemini-1.5-flash", 
            contents=prompt,
        )
        feedback = response.text
        
        # Attempt to get structured enum rating
        structured_eval = None
        try:
            # Attempt to get structured numeric score from the model 
             enum_response = client.models.generate_content(
                 model="gemini-1.5-flash",
                 contents=f"{prompt}\n\nBased on your evaluation above, what is the final score as a single digit from 1 to 5?",
                 generation_config=genai.types.GenerationConfig(
                     response_mime_type="text/plain", # Keep simple for just a digit
                     candidate_count=1,
                     max_output_tokens=5,
                     temperature=0.0
                 )
             )
             score_text = enum_response.text.strip()
             if score_text in [r.value for r in AwarenessRating]:
                 structured_eval = AwarenessRating(score_text)
             else:
                 print(f"Warning: Could not parse score '{score_text}' into enum.")

        except Exception as e_enum:
            print(f"Warning: Failed to get structured enum rating: {e_enum}")

        score_value = structured_eval.value if structured_eval else None
        score_name = structured_eval.name if structured_eval else None

        return feedback, score_value, score_name

    except Exception as e:
        print(f"Error during awareness evaluation: {e}")
        return f"Error evaluating awareness: {e}", None, None


# Simple goal completion tracking (e.g., 5/7 goals completed this week)
def calculate_goal_completion(user_data: Dict[str, Any]) -> float:
    """Track how consistently user meets wellness goals."""
    completed = sum(user_data.get("goals_completed", []))
    total = len(user_data.get("goals_completed", []))
    return round((completed / total) * 100, 2) if total else 0.0


@functools.cache
def detect_cognitive_shift_enhanced(text1: str, text2: str, habit_context: Optional[str] = None) -> str:
    """
    Compares two journal entries to detect emotional or cognitive shift.
    Uses Habit context if available to check integration or avoidance.
    """
    print(f"Detecting shift between texts (Habit context: {habit_context is not None})")
    prompt = SHIFT_PROMPT_ENHANCED.format(
        text1=text1,
        text2=text2,
        habit_context=habit_context if habit_context else "N/A"
    )
    try:
        response = client.models.generate_content(
            model="gemini-1.5-flash",
            contents=prompt
        )
        return response.text
    except Exception as e:
        print(f"Error during shift detection: {e}")
        return f"Error detecting shift: {e}"


# Using lru_cache for list inputs 
@functools.lru_cache(maxsize=128)
def generate_evolution_feedback_enhanced(entries_tuple: Tuple[str, ...], spiritual_context: Optional[str] = None) -> str:
    """
    Evaluates long-term growth across multiple entries.
    Considers emotional, cognitive and spiritual progression.
    """
    print(f"Generating evolution feedback for {len(entries_tuple)} entries (Spiritual context: {spiritual_context is not None})")
    # Format entries with index or date if available, otherwise just join
    # Using last 5 for brevity
    formatted_entries = "\n---\n".join([f"Entry {i+1}:\n{entry}" for i, entry in enumerate(reversed(entries_tuple[-5:]))])

    prompt = EVOLUTION_PROMPT_ENHANCED.format(
        entries=formatted_entries,
        spiritual_context=spiritual_context if spiritual_context else "N/A"
    )
    try:
        response = client.models.generate_content(
            model="gemini-1.5-flash", 
            contents=prompt
        )
        return response.text
    except Exception as e:
        print(f"Error during evolution feedback generation: {e}")
        return f"Error generating evolution feedback: {e}"
````

````python
def evaluate_user_growth_enhanced(
    entry_today: str,
    past_entries: List[str],
    goal_data: Dict[str, Any],
    # Optional context from the *last relevant agent interaction*
    last_cbt_output: Optional[str] = None,
    last_habit_output: Optional[str] = None,
    last_spiritual_output: Optional[str] = None
) -> Dict[str, Any]:
    """
    Master function that runs awareness, shift, evolution, and goal analysis.
    Each sub component integrates with CBT, Habit or Spiritual context where relevant.
    """
    print("\n Starting User Growth Evaluation ")
    # 1. Awareness check based on today's entry
    awareness_feedback, score, rating = eval_awareness_enhanced(
        entry_today,
        cbt_context=last_cbt_output
    )

    # 2. Detect shift from last entry, potentially using Habit context
    shift_feedback = "Not enough data to detect shift. Need at least one previous entry."
    if past_entries:
        # Use habit context if available and relevant to the shift period
        shift_feedback = detect_cognitive_shift_enhanced(
            past_entries[-1],
            entry_today,
            habit_context=last_habit_output # Pass context if relevant to this specific shift
        )

    # 3. Generate evolution feedback, potentially using Spiritual context
    # Convert list to tuple for lru_cache compatibility
    all_entries_tuple = tuple(past_entries + [entry_today])
    evolution_feedback = generate_evolution_feedback_enhanced(
        all_entries_tuple,
        spiritual_context=last_spiritual_output # Pass most recent spiritual context
    )

    # 4. Calculate goal completion 
    goal_completion_pct = calculate_goal_completion(goal_data)

    print(" Evaluation Complete ")
    return {
        "emotional_awareness": {
            "score": score,
            "rating": rating,
            "feedback": awareness_feedback 
        },
        "cognitive_shift": shift_feedback, 
        "goal_completion_pct": goal_completion_pct,
        "growth_summary": evolution_feedback 
    }
````
